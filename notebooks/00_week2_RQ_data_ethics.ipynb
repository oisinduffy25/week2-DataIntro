### 1. Where did the data come from?

The data was collected from high-school and early-college students (ages 13–21) using structured surveys. Voluntary & Anonymous: Participation was voluntary, and all identifying details were excluded to protect privacy. Time Window: The data was gathered over a period of 3 months across several urban and semi-urban schools. Consent: All participants (or their guardians) provided consent for anonymous use of data in research. 

### 2. Why was it originally collected?

This data was originally collected to combat and figure out the increasing use of social media and general smart phone usage within a younger population, more specifically teenagers.


### 3. Who is represented?

Young people in high schools and college students with ages ranging between 13 and 21.

### 4. Who might be missing?

While the study is about teen smartphone usage and addiction impact there are a lot of different ages groups missing that may indicate completely different data. Young adults ,around 22 - 35, may answer slightly or completely differently from the teens. Older generations as well ,50+, would also give different data.

In this dataset the sensitive information are these variables of each participant; Name, Age, Gender, Location, School Grade as well as other topics such as; Academic Performance, Social Interactions, and different variables about the participants mental health.

Our analysis could reveal hidden affects social media and daily phone usage has on teenagers and young children. This dataset only briefly goes into the topic of mental health and we plan to get a more detailed view on how these subjects health is affected.



### 7. Who could benefit from your findings?

Teenagers and parents/guardians can benefit from our findings as they can truly understand what could cause the addictions and construct possible solutions at home to either alleviate or control the addiction of smartphone usage.
